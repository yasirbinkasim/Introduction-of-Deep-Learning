```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create Neural Network
model = Sequential()

# Input + Hidden layer
model.add(Dense(
    units = 16,
    activation = "relu",
    input_shape = (2,)
))


# Hidden layer
model.add(Dense(
    units = 8,
    activation = "relu"
))
# Output layer
model.add(Dense(
    units = 1,
    activation = "sigmoid"
))

# display Model summary
model.summary()
```

# **Model compilation in Tensorflow & Keras**

Model compilation is the process of configuring a neural network by specifying the optimizer, loss function, and evaluation metrics before training.

Componenets:

**1. Optimizer -**
 An optimizer adjusts the weights of the neural networks to minimize the loss and amprove predictions accuracy

 . adam
 . SGD
 . RMSprop
 . Adagrad

 **2. Loss Function -**
 The loss function calculates the predictions error of the neural network.

 . binary_crosssentropy
 . categoriacl_crosssentropy
 . mean_squared_error(MSE)

 **3. Metrics -**
 Metrics evaluate how well the neural network is performing



```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create Model
model = Sequential()

# Input + Hidden layer
model.add(Dense(units = 16,activation = "relu",input_shape = (2,)))

# Hidden layer
model.add(Dense(units = 8,activation = "relu"))

# Output layer
model.add(Dense(units = 1,activation = "sigmoid"))

# compile Model
model.compile(
    optimizer = "adam",
    loss = "binary_crossentropy",
    metrics = ["accuracy"]
)

# model summary

model.summary()
```

# **Activation Functions**

An activation function determines the output of a neuron by transforming the weighted sum into a meaningful value.

Inputs (X) -> Weighted Sum -> Activation Function -> Output

**Types of Activation Functions**

1. Binary Step:-
 The binary step function gives only two outputs (0, 1) it is mainlu used in perceptron model

 2. Sigmoid Function:-
 It converts any input into a value between o and 1

 3. Tanh Function:-
 The tanh function output values betweeen -1 and +1

 4. ReLU (Rectified Linear Unit):-
  It is used activation function in deep learning

 5. Leaky ReLU:- Leaky Relu is an improved version of Relu. Instead of returning 0 for negative values, it returns a small negative value.

 6. Softmax:- Softmax is used for Multi-Class Classification. It converts outputs into probabilities whose total is 1.


# **Perceptron Model**

A perceptron is the simplest artificial Neural Ntwork (ANN) and is the building block of deep learning. it was introduced by Frank Roseblatt in 1958.

It is mainly used for Binary classification, where the output has only two classes:
1. Yes/No
2. Pass/ Fail
3. Spam/Not spam
4. Survived/ Not Survived

Inputs

X1      X2     X3
 |      |       |
 W1     W2     W3
  \      |     /
   \     |    /
    Weighted Sum
         |
       + Bias (b)
          |
    Activation Function
          |
        Output (0 or 1)

  

```
import numpy as np

# Inputs

hours = 8
attendance = 90

# weights

w1 = 0.6
w2 = 0.4

# Bias
bias = -4

# Weighted Sum

z = (hours * w1) + (attendance * w2) + bias

print("Weighted Sum:", z)

# Step Activation Function

if z >= 0:
  print("Prediction: Pass")
else:
  print("Prediction: Fail")
```
